```python
import pandas as pd
from matplotlib import pyplot as plt

pd.set_option('display.max_columns', None)
```

```python
listings = pd.read_csv('https://raw.githubusercontent.com/luminati-io/Airbnb-dataset-samples/refs/heads/main/airbnb-properties-information.csv')
listings.head()
```

```python
# Adjust data types
listings['category'] = listings['category'].astype('category')
bool_cols = ['availability', 'pets_allowed', 'is_supperhost', 'is_guest_favorite']
for col in bool_cols:
  listings[col] = listings[col].astype(bool)
```
