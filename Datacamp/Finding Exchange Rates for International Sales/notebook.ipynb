Exciting times! 

We've been running our very successful gadget webshop 'DataGadgets' for a few years and have recently expanded into new territories. While we've been focussed on the US market for the first five years of our existence, we now are shipping our cool data gadgets to the UK and Europe, too! But now our in-house built reporting has broken! Transactions don't only come in USD, but we're also receiving transactions in EUR and GPB. 

To better understand the volume of transactions being made, we should convert the non-USD transactions to USD and sum up the total. To do this, however, we'll need to use the proper exchange rates. 

In this project, we'll start with a CSV file containing all e-commerce transactions made on January 21st, but in their original currencies. Our job is to calculate the total sum in USD of all transactions so we know how much USD we sold on January 21st. To do this, we'll need to convert any non-USD transactions to USD using the exchange rate from January 21st, 2024. 

To get the exchange rates from January 21st, 2024, we'll rely on [VAT Comply rates API's](https://www.vatcomply.com/documentation#rates) public and free currency exchange API. We'll also use `pandas` to load the CSV file into a DataFrame and the `requests` package to make interacting with the API easier. 

We need to update the `orders` DataFrame so the final version has two new columns: `exchange_rate` and `amount_usd`. The final version should look as follows:

| `amount` | `currency` | `exchange_rate` | `amount_usd` |
|-|-|-|-|
| 43.75 | EUR | ... | ... |
| 385.5 | GBP | ... | ... |
| 495.5 | GBP | ... | ... |
| 117.99 | GBP | ... | ... |
| 624 | USD | ... | ... |

```python
# Import required packages/libraries
import pandas as pd
import requests

# Read the CSV file into a DataFrame
orders = pd.read_csv('data/orders-2024-01-21.csv')
orders.head()
```

```python
api_url = 'https://api.vatcomply.com/rates'
params = {
    'base': 'USD',
    'date': '2024-01-21'
}

response = requests.get(api_url, params=params)

if response.status_code == 200:
    exchange_rates = response.json()
else:
    exchange_rates = None

rates = exchange_rates['rates']
orders['exchange_rate'] = orders['currency'].map(rates)
orders['amount_usd'] = orders['exchange_rate'] * orders['amount']

orders.head()
```

```python
print(f"The total sum in USD of all transactions is {round(orders['amount_usd'].sum(),2)}")
```
