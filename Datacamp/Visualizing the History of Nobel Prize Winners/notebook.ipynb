The Nobel Prize has been among the most prestigious international awards since 1901. Each year, awards are bestowed in chemistry, literature, physics, physiology or medicine, economics, and peace. In addition to the honor, prestige, and substantial prize money, the recipient also gets a gold medal with an image of Alfred Nobel (1833 - 1896), who established the prize.

![](./images/Nobel_Prize.png)

The Nobel Foundation has made a dataset available of all prize winners from the outset of the awards from 1901 to 2023. The dataset used in this project is from the Nobel Prize API and is available in the `nobel.csv` file in the `data` folder.

In this project, we'll explore and answer several questions related to this prizewinning data. 

```python
# Loading in required libraries
import pandas as pd
import seaborn as sns
import numpy as np

df = pd.read_csv("data/nobel.csv")
```

## What is the most commonly awarded gender and birth country?

```python
top_gender = df['sex'].value_counts().index[0]
top_country = df['birth_country'].value_counts().index[0]
print(f"The most commonly awarded gender is {top_gender}, and the most awarded birth country is {top_country}")
```

## Which decade had the highest ratio of US-born Nobel Prize winners to total winners in all categories?

```python
import math

# Determining year range
first_year = df['year'].min()
last_year = df['year'].max()

# Determining year range
start_decade = math.floor(first_year / 10) * 10
end_decade = math.ceil(last_year / 10) * 10

# Create a separate column for decade by binning the year values
decades = [d for d in range(start_decade, end_decade+1,10)]
decade_bins = [e if i == 0 else e-1 for i, e in enumerate(decades)]
df['decade'] = pd.cut(df['year'], bins=decade_bins, labels=decades[:-1], include_lowest=True, precision=0)

# Split into US winners vs all
us_winners = df[df['birth_country'] == 'United States of America']
us_winner_counts = us_winners.groupby('decade', observed=True).count()['prize']
all_counts = df.groupby('decade', observed=True).count()['prize']

# Determine the ratio 
count_ratio = us_winner_counts / all_counts 
max_decade_usa = count_ratio.sort_values(ascending=False).index[0] # Sort & fetch the highest index label
print(f"The decade that had the highest ratio of US-born Nobel Prize winners to total winners is {max_decade_usa}")
```

## Which decade and Nobel Prize category combination had the highest proportion of female laureates?

```python
# Split into female winners vs all
female_winners = df[df['sex'] == 'Female']

# Determine the ratio
female_winner_counts = female_winners.groupby(['decade','category'], observed=True).count()['prize']
all_counts = df.groupby(['decade','category'], observed=True).count()['prize']
count_ratio = female_winner_counts / all_counts 
max_female_decade, max_female_category = count_ratio.sort_values(ascending=False).index[0] # Sort & fetch the highest index label
print(f"The highest proportion of female laureates occured in the decade {max_female_decade} in the category {max_female_category}")
```

## Who was the first woman to receive a Nobel Prize, and in what category?

```python
first_year = female_winners['year'].min()
first_win = female_winners[female_winners['year'] == first_year]
first_woman_name = first_win['full_name'].values[0]
first_woman_category = first_win['category'].values[0]
print(f"{first_woman_name} was the first woman to win a Nobel Prize in the {first_woman_category} category")
```

## Which individuals or organizations have won more than one Nobel Prize throughout the years?

```python
win_counts = df['full_name'].value_counts()
repeat_list = list(win_counts[win_counts > 1].index)
print(f"The individuals or organizations have won more than one Nobel Prize are:")
for n in repeat_list:
    print(f"- {n}")
```
